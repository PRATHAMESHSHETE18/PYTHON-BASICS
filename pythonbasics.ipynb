```python
# Q-6 Write a function to demonstrate the behavior of mutable and immutable arguments.
# Immutable
num = 10
new_num = num
new_num += 5  # Trying to change
print("Original num:", num)       # 10 (unchanged)
print("Modified new_num:", new_num)  # 15

# Mutable
my_list = [1, 2, 3]
new_list = my_list
new_list.append(4)  # Changing the list
print("Original my_list:", my_list)  # [1, 2, 3, 4] (changed)
print("Modified new_list:", new_list)  # [1, 2, 3, 4]
```

```python
# Q.13 - Write a Python program to find the sum of all even numbers between 1 and 50.

sum_even= 0

for num in range(1,51):
  if num % 2==0:
      sum_even += num

print("sum of all even numbers between 1 to 50 is :",sum_even)


```

```python
# Q.14 - Write a program to reverse a string using a while loop.
string=input("enter a string : ")
reversed_string=""
i=len(string)-1
while i>=0:
  reversed_string=reversed_string+string[i]
  i=i-1
print("reversed string",reversed_string)
```

```python
# Q-15 - Write a Python program to calculate the factorial of a number provided by the user using a while loop
num= int (input("enter a number :"))
factorial=1
n=num

while n>0:
  factorial=factorial*n
  n=n-1

print("factorial of number ", num, "is",factorial)
```
